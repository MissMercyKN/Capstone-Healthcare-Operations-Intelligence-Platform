# Healthcare Operations Intelligence Platform

## Notebook 3: Operational KPI Analysis

### Objective

This notebook transforms integrated operational data into executive performance indicators.

The KPIs developed here support:

- executive dashboard metrics
- operational monitoring
- congestion risk analysis
- predictive decision-making

The analysis combines healthcare demand, capacity, and workforce indicators.

```python
import pandas as pd
import numpy as np

from pathlib import Path

import plotly.express as px
```

```python
DATA_PATH = Path(
    "../data/processed/analytics_features_v3.csv"
)

df = pd.read_csv(DATA_PATH)

df.head()
```

KPI 1: Total Operational Observations

```python
total_observations = len(df)

total_observations
```

KPI 2: Average Waiting Time

```python
average_wait = (
    df["avg_wait_time"]
    .mean()
)

round(average_wait,2)
```

KPI 3: Average Treatment Duration

```python
average_treatment = (
    df["avg_treatment_time"]
    .mean()
)

round(average_treatment,2)
```

KPI 4: High Risk Rate

```python
high_risk_rate = (

    df["congestion_flag"]
    .mean()

) * 100


round(high_risk_rate,2)
```

KPI 5: Capacity Utilization KPI

```python
average_occupancy = (
    df["occupancy_rate"]
    .mean()
)

round(
    average_occupancy,
    2
)
```

KPI 6: Workforce Pressure KPI

```python
average_staff_pressure = (
    df["staff_ratio"]
    .mean()
)

round(
    average_staff_pressure,
    2
)
```

Department Performance Score

```python
department_kpi = (
    df.groupby("department")
    .agg(

        patient_volume=(
            "patient_volume",
            "mean"
        ),

        avg_wait_time=(
            "avg_wait_time",
            "mean"
        ),

        occupancy_rate=(
            "occupancy_rate",
            "mean"
        ),

        risk_score=(
            "risk_score",
            "mean"
        )

    )
    .reset_index()
)


department_kpi
```

Risk Visualization

```python
fig = px.bar(

    department_kpi,

    x="department",

    y="risk_score",

    title="Average Operational Risk Score by Department"

)


fig.show()
```

Executive KPI Table

```python
executive_kpis = pd.DataFrame({

    "Metric":[

        "Total Operational Records",

        "Average Waiting Time",

        "Average Occupancy",

        "High Risk Rate",

        "Average Risk Score"

    ],

    "Value":[

        total_observations,

        round(average_wait,2),

        round(average_occupancy,2),

        round(high_risk_rate,2),

        round(df["risk_score"].mean(),2)

    ]

})


executive_kpis
```

# # KPI Interpretation

The KPI framework converts operational data into measurable management indicators.

Key indicators include:

1. Patient volume to measure operational demand.

2. Average waiting time to measure service pressure.

3. Occupancy rate to measure capacity utilization.

4. Staff ratio to measure workforce pressure.

5. Risk score and congestion rate to identify departments requiring intervention.

These KPIs provide the foundation for executive dashboards and predictive operational analytics.
