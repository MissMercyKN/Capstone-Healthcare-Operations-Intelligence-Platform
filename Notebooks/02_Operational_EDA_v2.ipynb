# Healthcare Operations Intelligence Platform

## Notebook 2: Operational Exploratory Data Analysis

### Objective

This notebook analyzes healthcare operational patterns using the integrated analytics feature dataset.

The analysis focuses on:

- patient demand
- operational risk
- capacity pressure
- workforce pressure
- congestion drivers

The objective is to identify operational factors associated with increased congestion risk.

Section 2: Import Libraries

```python
import pandas as pd
import plotly.express as px
from pathlib import Path
```

```python
import os

print(os.getcwd())
```

Section 3: Load Data

```python
DATA_PATH = Path(
    "../data/processed/analytics_features_v3.csv"
)

df = pd.read_csv(DATA_PATH)

df.head()
```

Analysis 1: Department Workload

```python
department_volume = (
    df.groupby("department")
    ["patient_volume"]
    .mean()
    .reset_index()
)

department_volume
```

```python
fig = px.bar(
    department_volume,
    x="department",
    y="patient_volume",
    title="Average Patient Volume by Department"
)

fig.show()
```

Analysis 2: Waiting Time by Department

```python
waiting_analysis = (
    df.groupby("department")
    ["avg_wait_time"]
    .mean()
    .reset_index()
)

waiting_analysis
```

```python
fig = px.bar(
    waiting_analysis,
    x="department",
    y="avg_wait_time",
    title="Average Waiting Time by Department"
)

fig.show()
```

Analysis 3: Capacity Pressure Analysis

```python
capacity_analysis = (
    df.groupby("department")
    ["occupancy_rate"]
    .mean()
    .reset_index()
)

capacity_analysis
```

```python
fig = px.bar(
    capacity_analysis,
    x="department",
    y="occupancy_rate",
    title="Average Occupancy Rate by Department"
)

fig.show()
```

Analysis 4: Staffing Pressure Analysis

```python
staff_analysis = (
    df.groupby("department")
    ["staff_ratio"]
    .mean()
    .reset_index()
)

staff_analysis
```

```python
fig = px.bar(
    staff_analysis,
    x="department",
    y="staff_ratio",
    title="Staff Pressure by Department"
)

fig.show()
```

Analysis 5: Operational Risk Indicator

```python
risk_summary = (
    df.groupby("department")
    ["risk_score"]
    .mean()
    .reset_index()
)


risk_summary
```

```python
fig = px.bar(
    risk_summary,
    x="department",
    y="risk_score",
    title="Operational Risk Score by Department"
)

fig.show()
```

Risk Distribution

```python
risk_distribution = (
    df["congestion_flag"]
    .value_counts()
    .reset_index()
)

risk_distribution.columns = [
    "Risk Status",
    "Count"
]

risk_distribution
```

```python
fig = px.pie(
    risk_distribution,
    names="Risk Status",
    values="Count",
    title="Operational Congestion Risk Distribution"
)

fig.show()
```

# Operational Insights

The exploratory analysis identifies the main drivers of operational congestion.

Key findings:

1. Patient demand varies across departments, creating differences in workload pressure.

2. Capacity utilization contributes to operational risk, with higher occupancy increasing congestion likelihood.

3. Staffing pressure provides an additional indicator of operational strain.

4. The combined risk score provides a stronger representation of operational performance than waiting time alone.

5. These findings support the development of predictive models for proactive congestion management.
