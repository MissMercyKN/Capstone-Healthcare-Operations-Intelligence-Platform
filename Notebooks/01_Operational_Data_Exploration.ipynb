Risk Distribution

```python
risk_distribution = (
    df["congestion_flag"]
    .value_counts()
    .reset_index()
)

risk_distribution.columns = [
    "Risk Status",
    "Count"
]


risk_distribution
```

```python
fig = px.bar(
    risk_distribution,
    x="Risk Status",
    y="Count",
    title="Operational Risk Distribution"
)

fig.show()
```

Department Overview

```python
department_summary = (
    df.groupby("department")
    .agg(

        average_risk=(
            "risk_score",
            "mean"
        ),

        average_wait=(
            "avg_wait_time",
            "mean"
        ),

        average_occupancy=(
            "occupancy_rate",
            "mean"
        )

    )
    .reset_index()
)


department_summary
```

```python
fig = px.histogram(
    df,
    x="waiting_time",
    nbins=30,
    title="Patient Waiting Time Distribution"
)

fig.show()
```
